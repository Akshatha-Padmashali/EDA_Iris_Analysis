```python
column_names = [
    "SepalLength",
    "SepalWidth",
    "PetalLength",
    "PetalWidth",
    "Species"
]

df = pd.read_csv(
    "Iris Raw Dataset.csv",
    names=column_names
)

df.head()
```

```python
df["Species"].value_counts().plot(
    kind="bar"
)

plt.title("Species Distribution")

plt.xlabel("Species")

plt.ylabel("Count")

plt.show()
```

```python
df["SepalLength"].plot(
    kind="hist",
    bins=15
)

plt.title("Sepal Length Distribution")

plt.xlabel("Sepal Length")

plt.ylabel("Frequency")

plt.show()
```

```python
df["PetalLength"].plot(
    kind="hist",
    bins=15
)

plt.title("Petal Length Distribution")

plt.xlabel("Petal Length")

plt.ylabel("Frequency")

plt.show()
```

```python
plt.scatter(
    df["SepalLength"],
    df["PetalLength"]
)

plt.title(
    "Sepal Length vs Petal Length"
)

plt.xlabel("Sepal Length")

plt.ylabel("Petal Length")

plt.show()
```
